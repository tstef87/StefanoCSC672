3.) Explain how the Scikit-Learn ColumnTransformer and Pipeline streamline the data preparation process and prevent data leakage across preprocessing and training steps in your cloud pipeline.


Scikit-Learn’s ColumnTransformer and Pipeline combine preprocessing and modeling into one workflow. The ColumnTransformer handles missing values, scales numeric features, and encodes categorical features, while the Pipeline connects these steps to the model using fit() and predict(). This simplifies the process and prevents data leakage by fitting preprocessing steps only on training data and applying those same transformations to test data. I also removed CO(GT) from the features after creating the class labels to prevent target leakage.

6.) Based on your regularized regression models (Ridge versus Lasso), how did adjusting the regularization hyperparameter affect feature selection and coefficient shrinkage, and what operational environmental insights do these findings reveal?


Increasing alpha affected Ridge and Lasso differently. Ridge gradually reduced coefficients while keeping all features, with R2 only dropping from 0.8270 to 0.8246. Lasso eliminated less important features as alpha increased, reducing the number of features from 11 to 1 while R2 dropped from 0.826 to 0.065. An alpha of 0.01 provided a good balance, keeping 9 features with minimal accuracy loss. The strongest predictors were pollution-related measurements such as NOx(GT) and C6H6(GT), while temperature and humidity had less influence. These findings suggest that air quality monitoring could potentially use fewer sensors to reduce costs while maintaining reasonable prediction accuracy, as a five-feature model still achieved an R2 of approximately 0.79 compared to 0.83 for the full model.
