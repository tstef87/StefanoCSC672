1. How many principal components were required to retain 95% of the total variance compared to the original 4,096 dimensions? Based on your reconstruction plots in Task 3, what visual details are lost at lower values of k; e.g., k=10, and at what point does the reconstructed image become visually indistinguishable from the original?


Retaining 95% of the total variance required 123 principal components, compared to the original 4,096 dimensions, reducing the data to about 3% of its original size. At k = 10, the reconstructed faces appeared blurry, losing details such as facial features, hair texture, and expressions. As more components were added, the images became more clear and more recognizable. By k = 100, the reconstructions looked very similar to the originals, with only minor differences. Beyond this point, additional components provided little noticeable visual improvement. 


2. What do the earliest principal components (the first few "Eigenfaces") physically capture in the images? Why do later principal components appear increasingly like fine-grained pixel noise?

The first few eigenfaces captures lighting, shading, and overall facial structure. PC 1 explains 23.81% of the variance and mainly represents overall brightness, while PC 2 (13.99%) shows differences in lighting from left to right. PC 3 and PC 4 capture additional shading and head shape, while PC 5 through PC 8 begin showing more specific features like the eyes, mouth, glasses, and eyebrows. As the component number increases, the patterns become more detailed because PCA has already captured most of the major differences between faces. Later components explain much less variance and focus on smaller details, such as facial expressions, edges, and individual differences, which have less impact on the overall image.


3. Why is it critical to fit the PCA transformer strictly on the training split rather than on the entire dataset prior to splitting? What negative impacts occur if PCA is fitted on the full dataset before training downstream classifiers?

Fitting PCA only on training splits is important because it prevents data leakage and keeps the test set independent. PCA learns patterns from the data, so fitting it on the entire dataset would allow information from the test images to influence the principal components. This could lead to artificially higher accuracy and an unreliable evaluation of the model's performance on unseen data. By fitting PCA on the training set and applying that transformation to the test set, we simulate how the model would handle new images in real-world use. This is especially important with a small dataset of 400 images, where individual samples can have a greater influence on the results.


4. Compare the classification accuracy of Logistic Regression against Kernel SVM and the Ensemble Classifier on PCA-reduced features. Did non-linear kernels or ensemble methods yield substantial accuracy gains? Explain why classical classifiers (linear, non-linear, or ensemble) hit a performance ceiling when operating directly on flattened pixel vectors or global PCA projections.

Logistic Regression and Soft Voting achieved 97.5% accuracy, while Kernel SVM reached 95.0%. Neither the non-linear kernel nor the ensemble provided any accuracy improvement, with SVM performing slightly worse. These classifiers reach a performance ceiling because flattened pixel vectors lose spatial relationships between facial features, while PCA focuses on overall variation, such as lighting and shading, rather than individual identity. More complex classifiers cannot recover this missing spatial information. Additionally, the small dataset of only 8 training images per person limits how well models can learn patterns and increases the risk of overfitting.
